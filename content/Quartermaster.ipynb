```python
class Item:
    def __init__(self, name, facility, min_stock, max_stock, parts_list, tier, production_area, properties):
        self.__name = name
        self.__facility = facility
        self.__min_stock = min_stock
        self.__max_stock = max_stock
        self.__parts_list = parts_list  # List of tuples (part_name, quantity)
        self.__tier = tier
        self.__production_area = production_area
        self.__properties = properties  # Dictionary for additional properties
        self.__quantity = 0  # Current stock quantity (private)
        self.__quantity_log = []  # To keep a log of quantity updates
```

```python
    # Getter for name
    @property
    def name(self):
        return self.__name
```

```python
    # Getter for quantity
    @property
    def quantity(self):
        return self.__quantity

    # Setter for quantity with logging
    @quantity.setter
    def quantity(self, value):
        self.__log_quantity_update(self.__quantity, value)
        self.__quantity = value

    def __log_quantity_update(self, old_quantity, new_quantity):
        """Private method to log the quantity changes."""
        self.__quantity_log.append({
            'old_quantity': old_quantity,
            'new_quantity': new_quantity
        })

    def get_quantity_log(self):
        """Public method to access the quantity log."""
        return self.__quantity_log

    def is_below_min_stock(self):
        """Check if the item is below its minimum stock level."""
        return self.__quantity < self.__min_stock

    def is_above_max_stock(self):
        """Check if the item is above its maximum stock level."""
        return self.__quantity > self.__max_stock
```

```python
class Inventory:
    def __init__(self):
        self.__items = {}  # Private dictionary to store items

    def add_item(self, item):
        """Add an item to the inventory."""
        self.__items[item.name] = item

    def update_stock(self, item_name, quantity):
        """Update the stock of an item in the inventory."""
        if item_name in self.__items:
            self.__items[item_name].quantity = quantity
        else:
            print(f"Item {item_name} not found in inventory.")

    def calculate_parts_needed(self, item_name, quantity):
        """Calculate the parts needed to produce a certain quantity of an item."""
        if item_name not in self.__items:
            print(f"Item {item_name} not found in inventory.")
            return None

        parts_needed = {}

        def recursive_parts(item, qty):
            for part, part_qty in item.parts_list:
                total_part_qty = part_qty * qty

                # If the part itself has a parts list, recurse
                if self.__items[part].parts_list:
                    recursive_parts(self.__items[part], total_part_qty)
                else:
                    if part in parts_needed:
                        parts_needed[part] += total_part_qty
                    else:
                        parts_needed[part] = total_part_qty

        recursive_parts(self.__items[item_name], quantity)
        return parts_needed

    def generate_shopping_list_by_tier(self, tier):
        """Generate a shopping list for items of a specific tier."""
        shopping_list = {}
        for item in self.__items.values():
            if item.tier == tier:
                if item.is_below_min_stock():
                    shopping_list[item.name] = item.min_stock - item.quantity
                elif item.is_above_max_stock():
                    shopping_list[item.name] = item.quantity - item.max_stock

        return shopping_list
```

```python
# Example usage:

# Create an inventory object
inventory = Inventory()

# Add items to the inventory
inventory.add_item(Item('Widget A', 'Facility 1', 10, 50, [('Part X', 5), ('Part Y', 3)], 1, 'Area 1', {'color': 'blue'}))
inventory.add_item(Item('Part X', 'Facility 2', 20, 100, [], 1, 'Area 2', {'material': 'steel'}))
inventory.add_item(Item('Part Y', 'Facility 3', 30, 200, [('Part Z', 2)], 2, 'Area 3', {'size': 'small'}))
inventory.add_item(Item('Part Z', 'Facility 4', 5, 25, [], 2, 'Area 4', {'weight': 'light'}))

# Update stock with logging
inventory.update_stock('Widget A', 5)
inventory.update_stock('Part X', 50)
inventory.update_stock('Part Y', 15)
inventory.update_stock('Part Z', 10)

# Accessing the quantity log of Widget A
widget_a = inventory._Inventory__items['Widget A']
print("Quantity log for Widget A:", widget_a.get_quantity_log())
```
